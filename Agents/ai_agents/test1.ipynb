```python
import torch
import torch_xla
import torch_xla.core.xla_model as xm

device = torch_xla.device()

for _ in range(5):
    a = torch.randn(1000, 1000, device=device)
    b = torch.randn(1000, 1000, device=device)
    c = a @ b

    xm.mark_step()  # forces execution

print("Done")
```
