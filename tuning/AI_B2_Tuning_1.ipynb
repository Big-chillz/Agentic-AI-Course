```python
import torch #for core openrations, like tensors, numbers
import torch.nn as nn # to build neural networks
from transformers import AutoModelForCausalLM, AutoTokenizer # AutoModelForCasualLm - helps us load a pretrained language model, #AutoTokenizer - convert text to numbers (tensors)
```

```python
model = nn.Linear(in_features=2,out_features=1) # y = w1*x1 + w2*x2+b1
# weights = [w1,w2]
#bias = [b1]
```

```python
for name, param in model.named_parameters():
  print(f"{name}")
  print(param.data)
  print("Shape : ",param.shape)
  print("\n")
```

```python
a = model.parameters()
print(a)
```

```python
for param in model.parameters():
  print(param)
```

```python
total_params = sum(param.numel() for param in model.parameters())
print(total_params)
```

```python
#input
x = torch.tensor([[1.0, 2.0]])

output = model(x)

print("Input : ", x)
print("output : ",output)
```

```python
model_name = "distilgpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
hf_model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
for name, param in hf_model.named_parameters():
  print(f"{name}")
  print(param.data)
  print("Shape : ",param.shape)
  print("\n")
```

```python
total_params = sum(param.numel() for param in hf_model.parameters())
print(total_params)
```

```python
total_params_ref = sum(param.numel() for param in hf_model.parameters())
print(f"total parameters in {model} : {total_params_ref/1e6:.2f} Million")
```

```python
print(hf_model)
```

```python
layer = hf_model.transformer.h[0]
print(layer)
```

```python
weights = layer.attn.c_attn.weight
print("Shape", weights.shape)
print(weights)
```

```python
weights = layer.attn.c_attn.weight
print("Shape", weights.shape)
print(weights[:5,:5])
```

```python
input_text = "AI will change the world because"
inputs = tokenizer(input_text, return_tensors = "pt")

outputs = hf_model.generate(**inputs,max_length = 30)

print(tokenizer.decode(outputs[0]))
```
